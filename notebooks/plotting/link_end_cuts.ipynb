# Plot Link End Cuts

Plot the uncut buffer and the corridor (with end cuts) of example links, one row per link.

<br><br>

## 0. Setup

### 0.1. Imports and Config

```python
from splendor.config import STUDY_AREA, CORRIDOR_RADIUS
from splendor.io import load_network, load_raster
from splendor.plotting.landuse import plot_buffer_and_corridor
from splendor.paths import basemap05_main_types_path, ensure_parent, landuse_plot_path, network_path
from splendor.plotting.style import set_math_font

LEVEL = "links"
set_math_font()
```

### 0.2. Data

```python
_, edges = load_network(network_path(STUDY_AREA))
edges = edges.set_index("edge_id")
raster, transform, _ = load_raster(basemap05_main_types_path(STUDY_AREA))
```

<br><br>

## 1. End Cuts

### 1.1. Config

```python
EDGE_IDS = [295, 605]
DETAIL = "end_cuts"
OUT = landuse_plot_path(STUDY_AREA, LEVEL, CORRIDOR_RADIUS, DETAIL)
```

### 1.2. Plot

```python
for edge_id in EDGE_IDS:
    fig = plot_buffer_and_corridor(edges.geometry.loc[edge_id], CORRIDOR_RADIUS, raster, transform)
    out = landuse_plot_path(STUDY_AREA, LEVEL, CORRIDOR_RADIUS, f"end_cuts_{edge_id}")
    fig.savefig(ensure_parent(out), dpi=150)
    print(f"Saved {out}")
```
